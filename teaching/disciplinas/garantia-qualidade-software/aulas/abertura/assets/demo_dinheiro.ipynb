## 3. O parcelamento do `pedidos-api`

Esta é a função que está em produção no nosso exemplo. Ninguém escreveu nada errado aqui.

```python
def parcelas(total, n):
    valor = round(total / n, 2)
    return [valor] * n


for total, n in [(100.00, 3), (19.90, 7), (250.00, 6)]:
    p = parcelas(total, n)
    soma = round(sum(p), 2)
    print(f"  R$ {total:>7.2f} em {n}x = {n} x R$ {p[0]:.2f}"
          f" -> soma R$ {soma:.2f} | diferença R$ {round(soma - total, 2):+.2f}")
```

## 4. Um ano de operação

```python
perda = round(sum(parcelas(100.00, 3)), 2) - 100.00
print(f"  diferença por pedido : R$ {perda:+.2f}")
print(f"  x 500 pedidos/dia    : R$ {perda * 500:+.2f} por dia")
print(f"  x 365 dias           : R$ {perda * 500 * 365:+.2f} por ano")
```

### O mesmo teste, nos três casos

O teste acima cobre **um** caso. Este cobre os três — e acusa os três, inclusive
o de R$ 250,00 em 6x, aquele em que quem paga a mais é o **cliente**.

Um teste vale pelos casos que ele escolhe. É a sessão 7.

```python
falharam = []

for total, n in [(100.00, 3), (19.90, 7), (250.00, 6)]:
    try:
        assert sum(parcelas(total, n)) == total
    except AssertionError:
        falharam.append(f"R$ {total:.2f} em {n}x")

print("falharam:", falharam)
```

## "E como resolve?"

A resposta é a UC inteira. Só abrir esta parte **se a turma insistir** — 60 segundos.

Contar em **centavos inteiros** e distribuir o resto, em vez de repetir o mesmo valor.

```python
def parcelas_ok(total, n):
    centavos = round(total * 100)
    base, resto = divmod(centavos, n)
    return [(base + (1 if i < resto else 0)) / 100 for i in range(n)]


for total, n in [(100.00, 3), (19.90, 7), (250.00, 6)]:
    print(f"  R$ {total:>7.2f} em {n}x =", parcelas_ok(total, n))
```

### O mesmo teste, agora contra a versão corrigida

Passa. Nenhuma saída é o resultado esperado de um teste que passa.

```python
for total, n in [(100.00, 3), (19.90, 7), (250.00, 6)]:
    assert sum(parcelas_ok(total, n)) == total, f"R$ {total:.2f} em {n}x não fecha"

print("3 casos, nenhuma falha.")
```
